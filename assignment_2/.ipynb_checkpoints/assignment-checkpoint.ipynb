```python
import numpy as np

def f(x):
    """
    returns the value of the function f(x) = x^3 - x^2 - 1 at
    a given x
    """
    
    return x**3 - x**2 - 1

def df(x):
    """
    returns the value of the derivative of the function 
    f(x) = x^3 - x^2 - 1 evaluated at a given x
    """
    
    return 3*x**2-2*x

def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """
    Finds the root of a function using Newton's method given a known function,
    its derivative, and an initial position.
    
    Parameters:
    f: function
        the function whose root is being found
    df: function
        the derivative of f
    x0: float
        the initial guess for the root
    
    Returns a float if Newton's method is succesful in finding a root,
    or returns None when iteration within Newton's method exceeds 30 iterations.
    """
    
    i = 0
    initial = x0
    while i < 30:
        i = i+1
        if df(initial)==0:
            print(f"Iteration failed, df is zero at {initial}")
            print("blah")
            return None
        final = initial - f(initial)/df(initial)
        if np.abs(f(final)) < epsilon:
            print("blep")
            print(f"Found root in {i} iterations")
            break
        initial = final
        
    if i <= 30:
        return(final)
    print("Iteration failed")
    return None
```

```python
i = 0
initial = 100000000
epsilon = 1e-6
while i < 30:
    i = i+1
    print(i)
    print(df(initial))
    if df(initial)==0:
        print(f"Iteration failed, df is zero at {initial}")
    final = initial - f(initial)/df(initial)
    print(final)
    if np.abs(f(final)) < epsilon:
        print(f"Found root in {i} iterations")
    initial = final
        
if i <= 30:
    print(final)
print("Iteration failed")
```

```python
# Example 1, starting at 1 and 1e-6 as epsilon
print(newton(f, df, x0=1))
```

```python
# Example 2, starting at 50 and 1e-6 as epsilon
print(newton(f, df, 50)
```

```python
# Example 3, starting at 1 and 1e-8 as epsilon
print(newton(f, df, 1, epsilon=1e-8, max_iter=30))
```

```python
# Example 4, starting at 0 and 1e-6 as epsilon
print(newton(f, df, 0, epsilon=1e-8, max_iter=30))
```

```python
# Example 3, starting at 100000000 and 1e-8 as epsilon
print(newton(f, df, 100000000, epsilon=1e-8, max_iter=30))
```
