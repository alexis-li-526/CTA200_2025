```python
def if_diverge(z):
    """
    Takes an input iterable of size 2 and checks if the iteration
    z_i+1 = z_i ** 2 + c, z0 = 0 (where c complex number c composed of the
    first two inputs, where the the first input is the real component of c and the second
    input is the imaginary component of c) diverges. Defines divergence as the number of iterations
    taken to so that |z| > 2.
    
    If z diverges for the given input, returns the number of iterations taken so that |z| > 2.
    If z doesn't diverge in 1000 iterations, assumes z doesn't diverge and returns False.
    """
    
    # Check that z is indexable and has exactly two entries
    try:
        if len(z) != 2:
            raise ValueError("Input must have exactly two values: [real, imaginary].")
    except TypeError:
        raise TypeError("Input must be an iterable of size 2, like [real, imaginary].")

    # Check that both entries are numbers
    try:
        real = float(z[0])
        imag = float(z[1])
    except (TypeError, ValueError):
        raise TypeError("Both entries must be numbers.")

    
    c = z[0] + z[1]*1j
    i = 0
    z_i = 0
    # checks for divergence
    while i < 1000:
        i += 1
        z_f = z_i**2 + c
        if abs(z_f) > 2:
            return i
        z_i = z_f
    return False
```
